```python
path = "/content/drive/MyDrive/Library_Usage.csv"

import pandas as pd

df = pd.read_csv(path)

df.head()
```

```python
df.info()
```

```python
df["Patron Type Definition"].value_counts()
```

```python
df["Total Checkouts"].value_counts()
```

```python
import matplotlib.pyplot as plt

df["Year Patron Registered"].value_counts().plot(
    kind="pie",
)

plt.title("Patron Registration by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt

df["Year Patron Registered"].value_counts().plot(
    kind="bar",
)

plt.title("Patron Registration by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt

df["Year Patron Registered"].value_counts().plot(
    kind="bar",
    color="pink",
    figsize=(11,8),
)

plt.title("Patron Registration by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt

df["Circulation Active Month"].value_counts().plot(
    kind="bar",
    color="pink",
    figsize=(11,8),
)

plt.title("Library Circulation Activity by month")
plt.xlabel("# of Circulations")
plt.ylabel("Month")
plt.show()
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"]

df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    figsize=(10, 6),
    color="pink",
    edgecolor="black"
)

plt.title("Library Circulation Activity by month")
plt.xlabel("# of Circulations")
plt.ylabel("Month")
plt.show()
```

```python
df.isna().sum()
```

```python
df['Circulation Active Month Cleaned'] = df['Circulation Active Month'].fillna('Not Captured')
df.isna().sum()
```

```python
month_order = ["January","February","March","April","May","June","July","August","September","October","November","December", "Not Captured"]
month_counts = df["Circulation Active Month Cleaned"].value_counts().reindex(month_order)

month_counts.plot(
    kind="bar",
    color="pink",
    edgecolor="black",
    figsize=(10,6)
)

plt.title("Library Circulation Activity by Month", fontsize=16, weight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Number of Circulations", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
